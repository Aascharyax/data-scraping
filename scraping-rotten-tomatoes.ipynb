```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd

driver = webdriver.Chrome()

driver.get("https://www.rottentomatoes.com/browse/movies_at_home")

time.sleep(8)  # important: wait for JS load

# scroll a bit (loads more movies)
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(5)

# movie titles (more stable selector)
movies = driver.find_elements(By.CSS_SELECTOR, "a.js-tile-link")

titles = []

for m in movies:
    text = m.text.strip()
    if text:
        titles.append(text)

    if len(titles) == 50:
        break

driver.quit()

# remove duplicates / blanks
titles = list(dict.fromkeys([t for t in titles if t]))

df = pd.DataFrame({"Movie Title": titles})

df.to_excel("rotten_tomatoes_top50.xlsx", index=False)

print("Done! Movies scraped:", len(titles))
```
